```python
%load_ext autoreload
%autoreload 2

import sys
sys.path.append('..')

import cv2
import numpy as np
import matplotlib.pyplot as plt
import scipy
import random

import util
import blur
```

```python
# generate blur gaussian model
shape=127,127
n1,n2=shape[0],shape[1]
ax=0.015
bx=0.02
ay=0.015
by=0.03
gamma_x=1
gamma_y=1
gauss_pars=ax,bx,ay,by,gamma_x,gamma_y
gblur=blur.Gaussian(shape,gauss_pars,mode='approx')
```

```python
# sample points
n=shape[0]
t=8
lat=blur.Lattice(n,t,test_blur=gblur)
m1,m2=lat.m1,lat.m2

# grid 
lat.plot_grid(filename='../readme_visu/x_1.jpg')

# corrupted blur 
lat.plot_sampledpsf(filename='../readme_visu/y_1.jpg')
```

```python
lat.pca(r=10)
plt.plot(lat.D)
lat.plot_eigenpsf(2,5)
```

```python
plt.imshow(lat.mean_kernel.reshape(lat.m1,lat.m2))
```

```python
plt.imshow(lat.Cni)
```

```python
v=3

fig, (ax1, ax2) = plt.subplots(1, 2,figsize=(10,5))
ax1.imshow(lat.Cni[:,v].reshape(lat.t1,lat.t2))
ax1.axis('off')

ax2.imshow(lat.C[:,v].reshape(lat.n1,lat.n2))
ax2.axis('off')
```

```python
x=cv2.imread('../data/cameraman.tif',-1)
n1,n2=x.shape
alpha=0.498
n1=int(np.round(n1*alpha))
n2=int(np.round(n2*alpha))
print('new shape: ',(n1,n2))
x=cv2.resize(x,(n2,n1),interpolation=cv2.INTER_LINEAR)
x=util.normalize(x)

plt.imshow(x,cmap='gray');
plt.axis('off')
plt.savefig('../readme_visu/x_2.jpg',dpi=300, bbox_inches='tight',pad_inches=0)
```

```python
y1=gblur.forward(x,n_pool=24)
plt.imshow(y1,cmap='gray')
print(gblur.forward_timer)
```

```python
eblur=blur.Blur(shape,mode='approx',sampled_lattice=lat,w=gblur.w)
y2=eblur.forward(x,debug=True,n_pool=24)
plt.imshow(y2,cmap='gray')
plt.axis('off')
plt.savefig('../readme_visu/y_2.jpg',dpi=300, bbox_inches='tight',pad_inches=0)
print(eblur.forward_timer)
```

```python
_,ax = plt.subplots(1,2,figsize=(15,12))
ax[0].axis('off')
ax[1].axis('off')
ax[0].imshow(y1,cmap='gray')
ax[1].imshow(y2,cmap='gray')
```
